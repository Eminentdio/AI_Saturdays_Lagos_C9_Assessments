```python
import matplotlib.pyplot as plt
import random
import pandas as pd
from sklearn.datasets import load_breast_cancer

cancer = load_breast_cancer()
X, y = cancer.data, cancer.target
feature_names = cancer["feature_names"]
my_class_df = pd.DataFrame(X, columns=feature_names)
my_class_df["target"] = y

my_class_df.head()
```

```python
my_class_df.info
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaled = scaler.fit_transform(my_class_df[['worst fractal dimension', 'fractal dimension error']])

plt.scatter(scaled[:, 0][my_class_df['target'] == -1], scaled[:, 1][my_class_df['target'] == -1], color='red', label= 'negative')
plt.scatter(scaled[:, 0][my_class_df['target'] == 1], scaled[:, 1][my_class_df['target'] == 1], color='blue', label= 'positive')
plt.xlabel('fractal dimension error (Scaled)')
plt.ylabel('worst fractal dimension (Scaled)')
plt.legend()
plt.show()
```

```python
cancer["target"].tolist()
```

```python
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split

cancer_train_df, cancer_test_df = train_test_split(my_class_df, test_size=0.2, random_state=0)
X_cancer_train = cancer_train_df.drop(columns=["target"]).values
y_cancer_train = cancer_train_df["target"]
X_cancer_test = cancer_test_df.drop(columns=["target"]).values
y_cancer_test = cancer_test_df["target"]

#Normalization
scaler = MinMaxScaler()

X_cancer_train = scaler.fit_transform(X_cancer_train)
X_cancer_test = scaler.fit_transform(X_cancer_test)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

cancer_logReg = LogisticRegression()
cancer_logReg.fit(X_cancer_train, y_cancer_train)

cancer_train_accuracy = accuracy_score(y_cancer_train, cancer_logReg.predict(X_cancer_train))
cancer_test_accuracy = accuracy_score(y_cancer_test, cancer_logReg.predict(X_cancer_test))

print(
    f"Cancer Train accuracy is {cancer_train_accuracy}\n"
    f"Cancer Test accuracy is {cancer_test_accuracy}"
)
```

```python
from sklearn.svm import LinearSVC
svm = LinearSVC()
svm.fit(X_cancer_train, y_cancer_train)

cancer_train_accuracy = accuracy_score(y_cancer_train, svm.predict(X_cancer_train))
cancer_test_accuracy = accuracy_score(y_cancer_test, svm.predict(X_cancer_test))

print(
    f"Train accuracy is {cancer_train_accuracy}\n"
    f"Test accuracy is {cancer_test_accuracy}"
)
```
